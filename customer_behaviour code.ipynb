```python
cus['Review Rating']=cus.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
cus.info()
```

```python
cus.isnull().sum()
```

```python
cus.columns=cus.columns.str.lower()
cus.columns=cus.columns.str.replace(' ','_')
cus=cus.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
cus.columns
```

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
cus['age_group'] = pd.qcut(cus['age'], q=4, labels = labels)
```

```python
cus[['age','age_group']].head(10)
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

cus['purchase_frequency_days'] = cus['frequency_of_purchases'].map(frequency_mapping)
```

```python
cus[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
from sqlalchemy import create_engine

# Step 1: Connect to PostgreSQL
# Replace placeholders with your actual details

username = "postgres"          # default user
password = "1234"          # password you set during installation
host = "localhost"             # if running locally
port = "5432"                  # default PostgreSQL port
database = "customer_behaviour" # database you created in pgAdmin

engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

# Step 2: Load DataFrame into PostgreSQL

table_name = "customer"        # choose any table name

cus.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False
)

print(
    f"Data successfully loaded into table '{table_name}' "
    f"in database '{database}'."
)
```
